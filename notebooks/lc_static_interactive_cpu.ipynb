## Configure, Run and Results

Controls wrap with the available width. Applied voltage defaults to 0.9144 V.
Changing voltage uses the existing LCProp bias model on the next fresh run; it does
not continue a previous director state. **Restore qualified preset** restores all
inputs without changing an existing completed result. Modified configurations are
exploratory. Existing launch-sampling advisories are displayed above the preview.

Completed results show optical intensity and director reorientation XZ/YZ cuts through the
nearest sampled coordinates to the completed input beam center. Titles show the
actual sampled coordinates; z is horizontal. Width curves appear below. These are
retained slice-average intensities and director-minus-bias slices, not reconstructed optical fields.

## Use and export
Press **Run** once after reviewing the launch preview. Typical measured nonlinear
runtime: ~14 s local macOS, ~69 s hosted Colab; ~302 MiB observed child memory.
Plan at least 1 GiB free memory; no resource guarantee. The selected 120/180/300-second
between-slice budget applies. This synchronous first interface has no responsive
Stop control; kernel interruption is not a checkpoint. Completed results remain
separate from later drafts, failed runs and loaded requests.

Preview units are normalized intensity (1/µm²), **not physical irradiance**. A single
beam's normalized preview may remain unchanged when its power changes; physical power
still enters the canonical LC material calculation. Radius means 1/e field = 1/e²
intensity radius. Phase/group/orientation/wavelength are preserved from the preset.

**Export completed result** writes a new folder: canonical experiment, selected
scientific arrays, diagnostics, slice-position metadata, regenerated snapshot figure and SHA-256 manifest.
